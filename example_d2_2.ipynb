```python
def greet(name):
    return print(f"Hello, {name}!")

greet("Alice")

def evaluate_temperature(temp):
    if temp > 30:
        return print(f"{temp}°C? It's hot!")
    elif temp < 10:
        return print(f"{temp}°C? It's cold!")
    else:
        return print(f"{temp}°C? The temperature is moderate.")

evaluate_temperature(35)
evaluate_temperature(5)
evaluate_temperature(20)
```

```python
# function that return max and min from a list number
def max_min(numbers):
    if not numbers:
        return None, None
    return max(numbers), min(numbers)

# Example usage
numbers = [3, 7, 1, 9, 4]
maximum, minimum = max_min(numbers)
print(f"Maximum: {maximum}, Minimum: {minimum}")
```

```python
import re

states = [' Alabama ', 'Georgia!', 'Georgia', 'georgia', 'FlOrida', 'south carolina##', 'West virginia?']

for state in states:
    print(state)

# clean list text function
def clean_text(text):
    # Remove leading and trailing whitespace
    text = text.strip()
    # Remove special characters
    text = ''.join(char for char in text if char.isalnum() or char.isspace())
    # Convert to title case
    return text.title()

# clean list text function using regular expressions
def clean_text_regex(text):
    # Remove leading and trailing whitespace
    text = text.strip()
    # Remove special characters using regular expressions
    text = re.sub(r'[^a-zA-Z0-9\s]', '', text)
    # Convert to title case
    return text.title()

print("---------------------------\n")

# clean the states list
clean_states = [clean_text(state) for state in states]
for state in clean_states:
    print(state)

print("---------------------------\n")

# clean the states list using regular expressions
clean_states_regex = [clean_text_regex(state) for state in states]
for state in clean_states_regex:
    print(state)
```

```python
# Lambda functions
double = lambda x: x * 2
print(double(5))
```

```python
# Superior order functions (map, filter, sorted, reduce)
from functools import reduce
numbers = [1, 2, 3, 4, 5]

# Map example: double each number
doubled = list(map(lambda x: x * 2, numbers))
print(doubled)

# Filter example: keep only even numbers
evens = list(filter(lambda x: x % 2 == 0, numbers))
print(evens)

# Sorted example: sort numbers in descending order
sorted_numbers = sorted(numbers, reverse=True)
print(sorted_numbers)

# Reduce example: sum all numbers
total = reduce(lambda x, y: x + y, numbers)
print(total)
```

```python
words = ["elephant", "sun", "onion", "apple", "tree", "fish"]

# Ordena primero por longitud y, en caso de empate, alfabéticamente.
sorted_words = sorted(words, key=lambda word: (len(word), word))
print(sorted_words)
```
