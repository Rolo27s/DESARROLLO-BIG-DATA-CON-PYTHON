````python
import os
import sys

os.environ["PYSPARK_PYTHON"] = sys.executable
os.environ["PYSPARK_DRIVER_PYTHON"] = sys.executable

print("PYSPARK_PYTHON:", os.environ["PYSPARK_PYTHON"])
print("PYSPARK_DRIVER_PYTHON:", os.environ["PYSPARK_DRIVER_PYTHON"])
````

````python
from pyspark.sql import SparkSession

spark = (
    SparkSession.builder
    .master("local[*]")
    .config("spark.driver.host", "127.0.0.1")
    .config("spark.driver.bindAddress", "127.0.0.1")
    .getOrCreate()
)

sc = spark.sparkContext
print("SparkContext created successfully:", sc)
print("Spark version:", sc.version)
````

````python
sc.parallelize([1, 2, 3]).collect()
print("RDD created and collected successfully.")
````

Línea de codigo en consola para traer el fichero csv
```powershell

wget -O 50_Startups.csv https://raw.githubusercontent.com/gakudo-ai/open-datasets/refs/heads/main/50_Startups.csv
```

````python
rdd = sc.textFile("50_Startups.csv")
rdd_header = rdd.first()
rdd_data = rdd.filter(lambda row: row != rdd_header)

rdd_data.take(5)
````

````python
# Filtrar startups de Nueva York
rdd_ny = rdd_data.filter(lambda row: "New York" in row)
# Mostrar las primeras 5 filas de startups de Nueva York
print("First 5 startups in New York:", rdd_ny.take(5))

print("Startups count in New York:", rdd_ny.count())
````

````python
# Cerrar sesion de spark
spark.stop()
print("Spark session stopped successfully.")
````
