```python
def mi_funcion(x, y, z=1.5, w=44):
    if z > 1:
        return x + y + z - w
    return x + y - w

print(mi_funcion(2, 3))
print(mi_funcion(2, 3, z=22))
print(mi_funcion(2, 3, w=20))
print(mi_funcion(2, 3, 55, 20))
```
