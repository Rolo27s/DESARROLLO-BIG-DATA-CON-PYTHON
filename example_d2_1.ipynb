```python
temperature = 25

if temperature > 30:
    print("It's hot")
elif temperature < 10:
    print("It's cold")
else:
    print("The weather is pleasant")

# Using match-case to handle temperature conditions
match temperature:
    case _ if temperature > 30:
        print("It's hot")
    case _ if temperature < 10:
        print("It's cold")
    case _:
        print("The weather is pleasant")

```

```python
# Logical operators example (AND, OR, XOR, NOR, ...)
a = True
b = False

# AND operator
print("AND:", a and b)

# OR operator
print("OR:", a or b)

# XOR operator (using != for logical XOR)
print("XOR:", a != b)

# NOR operator (using not with OR)
print("NOR:", not (a or b))

# NAND operator (using not with AND)
print("NAND:", not (a and b))

# XNOR operator (using not with XOR)
print("XNOR:", not (a != b))
```

```python
# Ternary operator example
temperature = 25
weather = "hot" if temperature > 30 else "cold" if temperature < 10 else "pleasant"
print("The weather is", weather)

age = 19
message = "You are an adult." if age >= 18 else "You are a minor."
print(message)
```

```python
# Value list dinamic
values = [i for i in range(10)]
print(values)

# Same but only even numbers
even_values = [i for i in range(10) if i % 2 == 0]
print(even_values)
```

```python
# print list of values
fruit_list = ["apple", "banana", "cherry"]
print(fruit_list)

for fruit in fruit_list:
    print(fruit)

for i, fruit in enumerate(fruit_list):
    print(i, fruit)

for i in range(len(fruit_list)):
    print(i, fruit_list[i])
```

```python
# print set (no order collection)
fruit_set = {"apple", "banana", "cherry"}
print(fruit_set)

for fruit in fruit_set:
    print(fruit)

for i, fruit in enumerate(fruit_set):
    print(i, fruit)

for i in range(len(fruit_set)):
    print(i, list(fruit_set)[i])
```

```python
# print a dict
my_dict = {"name": "Alice", "age": 25, "city": "New York"}
print(my_dict)

for key, value in my_dict.items():
    print(key, value)

for index, (key, value) in enumerate(my_dict.items()):
    print(index, key, value)
```

```python
# Merge two list (name + age)
names = ["Alice", "Bob", "Charlie"]
ages = [25, 30, 35]
merged_list = list(zip(names, ages))
print(merged_list)

for name, age in zip(names, ages):
    print(name, age)
```

```python
try:
    num = int(input("Enter a number: "))
    result = 10 / num
    print("Result:", result)
except ValueError:
    print("Invalid input. Please enter a valid number.")
except ZeroDivisionError:
    print("Cannot divide by zero.")
finally:
    print("Execution completed.")
```
