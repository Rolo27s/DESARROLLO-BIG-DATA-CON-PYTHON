# Pandas: selección de datos

Esta guía muestra las formas más habituales de seleccionar filas y columnas de un `DataFrame` con pandas.

Trabajaremos con un conjunto de datos pequeño para que cada operación sea fácil de interpretar.

```python
import pandas as pd

df = pd.DataFrame({
    "nombre": ["Ana", "Luis", "Marta", "Pablo"],
    "edad": [25, 32, 28, 41],
    "ciudad": ["Madrid", "Barcelona", "Madrid", "Sevilla"],
    "salario": [28000, 35000, 31000, 42000]
})

df
```

## Seleccionar columnas

Una columna devuelve una `Series`; varias columnas devuelven otro `DataFrame`.

```python
# Una columna: devuelve una Serie
df["nombre"]

# Varias columnas: devuelve un DataFrame
df[["nombre", "ciudad"]]
```

## Seleccionar filas: `iloc` y `loc`

`iloc` selecciona por posición numérica. `loc` selecciona por etiquetas del índice y permite combinar condiciones con columnas.

```python
# Primera fila
df.iloc[0]

# Filas 1 y 2
df.iloc[1:3]

# Filas 0 y 2 con columnas concretas
df.loc[[0, 2], ["nombre", "salario"]]
```

## Filtrar filas

Los filtros se construyen con una condición booleana. Para combinar condiciones se usan `&` para AND y `|` para OR, siempre con paréntesis.

```python
# Personas mayores de 30 años
df[df["edad"] > 30]

# Personas de Madrid
df[df["ciudad"] == "Madrid"]

# Varias condiciones
df[(df["edad"] >= 28) & (df["salario"] > 30000)]

# Valores incluidos en una lista
df[df["ciudad"].isin(["Madrid", "Sevilla"])]

# Filtrar y seleccionar columnas a la vez
df.loc[df["edad"] > 30, ["nombre", "salario"]]

# Alternativa usando query
df.query("edad > 30 and salario >= 35000")
```

## Valores nulos y resumen

`isna()` permite localizar valores ausentes. `dropna()` elimina las filas que contienen algún valor nulo.

```python
# Añadir un valor nulo solo para demostrar la selección
df_con_nulos = df.copy()
df_con_nulos.loc[1, "salario"] = None

# Filas con algún valor nulo
df_con_nulos[df_con_nulos.isna().any(axis=1)]

# Filas sin ningún valor nulo
df_sin_nulos = df_con_nulos.dropna()
df_sin_nulos
```

## Regla rápida

- Usa `df["columna"]` para una columna.
- Usa `df[["col1", "col2"]]` para varias columnas.
- Usa `iloc` cuando selecciones por posición.
- Usa `loc` cuando selecciones por etiquetas o condiciones.
- Usa `query()` cuando quieras expresar filtros de forma legible.
